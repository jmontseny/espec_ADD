## 1. Exploración y gestión de una API de laboratorio

Para familiarizarte con el consumo de APIs REST, comenzarás trabajando con una API de pruebas que simula el comportamiento de un servicio real.

El equipo quiere comprobar que eres capaz de consultar diferentes recursos, interpretar las respuestas del servidor y realizar operaciones básicas sobre los datos.

Durante esta fase deberás trabajar con información relacionada con:

Además de recuperar información, se plantea un escenario en el que tendrás que gestionar diferentes situaciones habituales al trabajar con APIs:

- Comprobar que una petición se ha realizado correctamente.
- Identificar qué ocurre cuando se solicita un recurso que no existe.
- Crear una nueva publicación.
- Modificar parcialmente información existente.
- Eliminar una publicación.

### Objetivo de negocio

Demostrar que eres capaz de interactuar con un servicio externo y distinguir entre operaciones de consulta y operaciones de modificación, así como interpretar correctamente las respuestas del servidor.

```python
# Creamos los endpoints que vamos a utilizar
endpoints = ["posts", "users", "todos"]

data = {}

for endpoint in endpoints:
    url = "https://jsonplaceholder.typicode.com/" + endpoint

    response = requests.get(url)
    # mostramos el código de estado de cada petición
    print(f'Status code response for /{endpoint}: {response.status_code}')

    data[endpoint] = response.json()
```

```python
# Mostramos la cantidad total de cada recurso

for endpoint in endpoints:
    print(f'Elements in data["{endpoint}"]: {len(data[endpoint])}')
```

```python
# Hacemos una petición a una publicación inexistente
response = requests.get("https://jsonplaceholder.typicode.com/publicacioninexistente")

response.status_code
```

```python
# Analizamos cómo están compuestos los posts
data['posts'][0]
```

```python
# Hacemos una petición POST para crear una nueva publicación ficticia
url = "https://jsonplaceholder.typicode.com/posts"

new_post = {
    "userId": random.randint(1, 10),
    "id": data['posts'][-1]['id']+1,
    "title": "titulum inventatum",
    "body": "Lorem ipsum dolor sit amet, consectetur adipiscing elit, sed do eiusmod tempor incididunt ut labore et dolore magna aliqua."
}

response = requests.post(url, json=new_post)

# Mostramos la respuesta JSON
print(response.json())

# Mostramos el código de estado
print(response.status_code)
print(response)
```

```python
# Hacemos una petición PATCH para modifficar parcialmente una publicación existente
url = "https://jsonplaceholder.typicode.com/posts/1"

data = {
    "title": "novum titulum"
}

response = requests.patch(
    url,
    json=data
)

# Mostramos la respuesta JSON
print(response.json())

# Mostramos el código de estado
print(response.status_code)
```

```python
# Hacemos una petición DELETE para borrar una publicación existente

response = requests.delete(url)

# Mostramos la respuesta JSON
print(response.json())

# Mostramos el código de estado
print(response.status_code)
```

## Integración con una API pública real

Una vez familiarizado con una API de laboratorio, el equipo quiere dar un paso más y trabajar con una fuente de datos pública real.

Deberás seleccionar una API pública que pueda aportar información potencialmente útil para un proyecto de análisis de datos y estudiar cómo está estructurada antes de comenzar a consumirla.

El objetivo será identificar qué información proporciona el servicio, qué recursos están disponibles y qué posibilidades ofrece para limitar o personalizar las consultas.

Posteriormente realizarás una primera extracción de datos y transformarás la respuesta obtenida en un DataFrame de pandas, dejándolo preparado para posteriores procesos de análisis y exploración.

### Objetivo de negocio

Evaluar la capacidad para incorporar una fuente de datos externa al flujo de trabajo de análisis, entendiendo previamente qué información proporciona y cómo puede utilizarse.

Esta fase representa un escenario habitual en proyectos de datos: antes de incorporar una API a un proceso ETL, el analista debe conocer la estructura, posibilidades y limitaciones de la fuente.

```python
# Utilizaremos la API de Frankfurter

url = "https://api.frankfurter.dev/v2/rates"
response = requests.get(url)

print(response)
response.json()
```

Revisando la sección de endpoints, podemos ver los tipos de cambio actuales con el endpoint:

/v2/rates

También podemos cambiar la divisa (currency) base con el endpoint:

/v2/rates?base={currency}

Podemos filtrar tipos de cambio para una fecha concreta con:

/v2/rates?date={YYYY-MM-DD}

Podemos obtener una serie temporal entre dos fechas con:

/v2/rates?from={YYYY-MM-DD}&to={YYYY-MM-DD}

```python
# Hacemos una petición sencilla

response = requests.get(
    url,
    params={
        "base": "EUR",
        "quotes": "USD,GBP"
    }
)

print("Status code:", response.status_code)

data = response.json()

print(data)
```

```python
# Imprimimos algunos de los campos de la respuesta JSON

for currency in data:
    print(
        f"Date: {currency['date']} | "
        f"From: {currency['base']} | "
        f"To: {currency['quote']} | "
        f"Rate of change: {currency['rate']}"
    )
```

```python
# Lo convertimos a DataFrame y lo mostramos
df_rates = pd.DataFrame(data)

df_rates
```

## Explotación de Open Data Barcelona

El Ayuntamiento de Barcelona dispone de un amplio catálogo de datos abiertos que puede utilizarse para desarrollar análisis sobre diferentes aspectos de la ciudad.

El equipo quiere incorporar una fuente de Open Data a sus procesos de análisis y necesita identificar un conjunto de datos que pueda resultar relevante para un futuro proyecto.

Deberás explorar el catálogo de Open Data BCN, localizar un dataset de interés y comprobar que dispone de información estructurada que pueda ser consumida mediante su API.

Una vez seleccionado el conjunto de datos, realizarás una extracción de información suficiente para trabajar con ella y la transformarás en un DataFrame de pandas.

Finalmente, el conjunto de datos obtenido deberá quedar almacenado en formato CSV para poder reutilizarlo posteriormente en otros procesos de análisis.

### Objetivo de negocio

Completar un pequeño flujo de adquisición y preparación de datos procedentes de una fuente pública real, dejando el dataset preparado para incorporarlo a un futuro proceso de análisis.

El resultado final debe permitir pasar de una fuente externa de datos abiertos a un fichero estructurado y reutilizable:

Open Data BCN → API → datos estructurados → DataFrame → CSV

```python
# Creamos la URL
base_url = "https://opendata-ajuntament.barcelona.cat/data/api/action/"

url = base_url + "package_search"

# Buscamos un dataset con package_search relacionado con la electricidad
params = {
    "q": "electric"
}

response = requests.get(url, params=params)

print(response.status_code)

data = response.json()
data
```

```python
for dataset in data["result"]["results"]:
    print(dataset["name"])
    print(dataset["title"])
    print()
```

```python
# Nos quedamos con el primer dataset
dataset_name = data["result"]["results"][0]["name"]

url = base_url + "package_show"

params = {
    "id": dataset_name
}

response = requests.get(url, params=params)

dataset = response.json()
dataset["result"]
```

```python
# Buscamos los recursos diponibles
dataset["result"]["resources"]
```

```python
for resource in dataset["result"]["resources"]:
    print("Dataset name:", resource.get("name"))
    print("Format:", resource.get("format"))
    print("Resource ID:", resource.get("id"))
    print("URL:", resource.get("url"))
    print()
```

```python
# Nos quedamos sólo con aquellos recursos cuyo formato sea CSV o JSON
for resource in dataset["result"]["resources"]:
    
    if resource.get("format", "").upper() in ["CSV", "JSON"]:
        print(resource["name"])
        print(resource["format"])
        print(resource["id"])
        print()
```

```python
# Con datastore_search realizamos una consulta para recuperar al menos 100 registros de un resource_id

url = base_url + "datastore_search"

for resource in dataset["result"]["resources"]:
    
    if resource.get("format", "").upper() in ["CSV", "JSON"]:

        resource_id = resource["id"]

        params = {
            "resource_id": resource_id,
            "limit": 100
        }

        response = requests.get(url, params=params)

        if response.status_code == 200:
            print(resource["name"])
            print(resource["format"])
            data = response.json()
            break
```

```python
data["result"]["records"]
```

```python
# Convertimos el resultado a un DataFrame
df = pd.DataFrame(data["result"]["records"])

df.head()
```

```python
# Convertimos el DataFrame a un fichero .csv

df.to_csv("dataset_2024_consum_electricitat_bcn.csv", index=False)
```
